# 01 — Acquire or reuse a NEON HDF5

Use this focused notebook when you want only the acquisition step from the full orchestrator. It keeps the same identifiers as `Raster_processing.ipynb`, calls the pipeline's restart-safe download stage, and checks the concrete HDF5 path afterward.

## 1. Configure the NEON request

Copy these values from a working full-pipeline configuration. The flightline stem is the canonical filename stem without `.h5`. NEON requires a Data API token for downloads; export `NEON_API_TOKEN` (or `NEON_TOKEN`) in the environment before starting Jupyter. Do not paste a token into this notebook or commit it.

```python
from pathlib import Path

from spectralbridge.pipelines.pipeline import stage_download_h5

RUN = False
base_folder = Path("outputs/neon_notebook")
site_code = "NIWO"
product_code = "DP1.30006.001"
year_month = "2023-08"
flight_stem = "NEON_D13_NIWO_DP1_L019-1_20230815_directional_reflectance"
expected_h5 = base_folder / f"{flight_stem}.h5"
```

## 2. Download or reuse

A non-empty valid local file is reused. Leave `RUN = False` while checking the resolved path; setting it to `True` may make a network request.

```python
print(f"Expected HDF5: {expected_h5.resolve()}")
h5_path = expected_h5
if RUN:
    h5_path = stage_download_h5(
        base_folder,
        site_code,
        year_month,
        product_code,
        flight_stem,
    )
else:
    print("No network request made. Set RUN = True after checking the identifiers.")
```

## 3. Check the downloaded file

The next stage expects this exact path. Size is a quick diagnostic; downstream HDF5 validation remains authoritative.

```python
print(f"Exists: {h5_path.exists()}")
if h5_path.exists():
    print(f"Size: {h5_path.stat().st_size:,} bytes")
    print(f"Ready for notebook 02: {h5_path}")
else:
    print("HDF5 is not present yet. Run this stage or return to notebook 00.")
```
