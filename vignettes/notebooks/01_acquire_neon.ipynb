# 01 — Acquire or reuse a NEON HDF5

This stage ensures the canonical `<base_folder>/<flight_stem>.h5` exists. It reuses a non-empty local file and downloads only when needed.

```python
from pathlib import Path
from spectralbridge.pipelines.pipeline import stage_download_h5

RUN = False
BASE_FOLDER = Path("outputs/neon_notebook")
SITE_CODE = "NIWO"
YEAR_MONTH = "2023-08"
PRODUCT_CODE = "DP1.30006.001"
FLIGHT_STEM = "NEON_D13_NIWO_DP1_L019-1_20230815_directional_reflectance"
```

```python
expected = BASE_FOLDER / f"{FLIGHT_STEM}.h5"
print(f"Expected HDF5: {expected.resolve()}")
if RUN:
    h5_path = stage_download_h5(BASE_FOLDER, SITE_CODE, YEAR_MONTH, PRODUCT_CODE, FLIGHT_STEM)
    print(f"Ready: {h5_path} ({h5_path.stat().st_size:,} bytes)")
else:
    print("No network request made. Set RUN = True after checking the identifiers.")
```
